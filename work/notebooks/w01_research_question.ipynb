The decision is: **which content items should be reviewed first for refresh, expansion, protection, pruning, or monitoring?** A content or SEO reviewer acts on the ranked queue by checking the evidence and choosing an appropriate next action. ## 3. Quick look at the data (2-3 real numbers)

The starter file is a 30,000-row anonymized snapshot covering 32 pseudonymized clients. In it, 16,726 rows have at least 500 impressions in the trailing 90 days, so there is a meaningful visible-page population for a review queue. The current-window proxy marks 16,262 rows as `down` (54.2%), which is large enough to inspect but not a future outcome. I will use this quick look to justify further signal checks, not to claim that a refresh will cause recovery.

```python
from pathlib import Path
import pandas as pd

candidate_paths = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../../data/raw/content_refresh_anonymized.csv"),
]
data_path = next(path for path in candidate_paths if path.exists())
df = pd.read_csv(data_path)

visible_pages = df["impressions_90d"] >= 500
down_proxy = df["trend_direction"].eq("down")

print(f"Rows: {len(df):,}")
print(f"Pseudonymized clients: {df['client_id'].nunique():,}")
print(f"Pages with at least 500 impressions: {visible_pages.sum():,}")
print(f"Rows marked down by the current-window proxy: {down_proxy.sum():,} ({down_proxy.mean():.1%})")

assert len(df) == 30_000
assert df["client_id"].nunique() == 32
assert visible_pages.sum() == 16_726
assert down_proxy.sum() == 16_262
```
